## Classification using ANN
### 1- an MLP   ,   2- A Deep CNN
### Dataset is Diabetes disease: datasets_228_482_diabetes.csv
### 

```python
# Regression Example With HDB Dataset
import pandas as pd
from pandas import read_csv, read_excel
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.layers import Dense , Flatten
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D , MaxPooling1D , Dropout
from tensorflow.keras.wrappers.scikit_learn import KerasRegressor
from sklearn.model_selection import train_test_split 
from sklearn.preprocessing import StandardScaler
import sklearn.metrics as skm
import numpy as np
import matplotlib.pyplot as plt  
from sklearn.metrics import classification_report, confusion_matrix
```

```python

# load dataset
dataframe = read_csv("datasets_228_482_diabetes.csv")
dataset = dataframe.values

# split into input (X) and output (Y) variables, 5 input/independent vars, output is the HDB 
# resale price in S$
X = dataset[:,0:8]
Y = dataset[:,8]
```

```python
#inputs/output vars, shape of the dataset, description of that
print(dataframe)
print('\n==============\n', dataset.shape)
print( '\n==============\n' , dataframe.describe() )
```

```python
# split the train / test subsets, 90% for training, repeatable
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=110)
```

```python
# normalization using normal scaler, result is having each feature in mean=0 and std=1
# print the test input features subset before normalization
print(X_test)
# normalization, both test and train features
scaler = StandardScaler()
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
# print the test input features after normalization
print("\n-------------------\n", X_test)
```

## Part1: MLP
#### structure: 5x4x10x1

```python
# define base model as a function, we compile it inside that function too.
# loss function and accuracy metric both are MSE, training algo is Adam
#
def baseline_model():
    # create model
    model = Sequential()
    model.add(Dense(4, input_dim=8, kernel_initializer='normal', activation='relu'))
    model.add(Dense(10, kernel_initializer='normal', activation='relu'))
    model.add(Dense(1, kernel_initializer='normal', activation='sigmoid'))
    # Compile model
    model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
    
    return model
```

```python
# train our baseline regressor model
estimator1 = baseline_model()
estimator1.fit(X_train, Y_train, epochs=800 , verbose=1 , batch_size=15)
```

```python
# lets see what the estimated prices for HDBs in the test subset are
#
yp = estimator1.predict(X_test, verbose=1)
print('Detected Classes: \n', yp)
```

```python
# Training performance
#
yprd1 = estimator1.predict(X_train, verbose=1)
yprd1 = np.round(yprd1)
print(confusion_matrix(Y_train, yprd1))
print('\n\n', classification_report(Y_train, yprd1))
```

```python
# Testing performance
#
yprd1 = estimator1.predict(X_test, verbose=1)
yprd1 = np.round(yprd1)
print(confusion_matrix(Y_test, yprd1))
print('\n\n', classification_report(Y_test, yprd1))
```

## Part 2: Deep Learning

```python
# define the deep convolutional model
def deep_model():
    # create model
    model = Sequential()
    model.add(Conv1D(16, 2, activation="relu", input_shape=(8,1)))
    model.add(Conv1D(12, 2, activation="relu"))
    #model.add(Dense(16, activation="relu"))
    #model.add(MaxPooling1D())
    model.add(Dropout(0.25))
    model.add(Flatten())
    
    model.add(Dense(1, kernel_initializer='normal', activation='sigmoid'))
    # Compile model
    model.compile(loss='binary_crossentropy', optimizer='adam')
    return model
```

```python
# train the DCNN model
estimator2 = deep_model()
estimator2.fit(X_train, Y_train, epochs=1000 , verbose=1 , batch_size=15)
```

```python
# Training performance
#
yprd1 = estimator2.predict(X_train, verbose=1)
yprd1 = np.round(yprd1)
print(confusion_matrix(Y_train, yprd1))
print('\n\n', classification_report(Y_train, yprd1))
```

```python
# Testing performance
#
yprd1 = estimator2.predict(X_test, verbose=1)
yprd1 = np.round(yprd1)
print(confusion_matrix(Y_test, yprd1))
print('\n\n', classification_report(Y_test, yprd1))
```
