```python
import os
import logging
import sys
import numpy as np
import pandas as pd

root_logger = logging.getLogger()
for handler in root_logger.handlers[:]:
    root_logger.removeHandler(handler)

file_handler = logging.FileHandler("pipeline.log", mode="w")
stream_handler = logging.StreamHandler()
log_formatter = logging.Formatter('%(asctime)s - %(levelname)s - %(message)s')

file_handler.setFormatter(log_formatter)
stream_handler.setFormatter(log_formatter)

logger = logging.getLogger(__name__)
logger.setLevel(logging.INFO)
logger.addHandler(file_handler)
logger.addHandler(stream_handler)

logger.info("Starting the traffic data pipeline.")

def load_csv_data(file_name):
    logger.info("Attempting to load the CSV file.")
    if not os.path.exists(file_name):
        logger.error("Error: The CSV file was not found in this folder.")
        return None
    try:
        data = pd.read_csv(file_name)
        rows, cols = data.shape
        logger.info(f"Success: CSV file loaded into memory. Loaded {rows} rows and {cols} columns.")
        return data
    except pd.errors.ParserError:
        logger.error("Error: The CSV file formatting is corrupted.")
        return None
    except Exception as e:
        logger.error(f"An unexpected error happened: {e}")
        return None

def validate_schema(df):
    if df is None:
        return False
    logger.info("Validating the data schema columns.")
    expected_columns = [
        'holiday', 'temp', 'rain_1h', 'snow_1h', 'clouds_all', 
        'weather_main', 'weather_description', 'date_time', 'traffic_volume'
    ]
    missing_columns = []
    for col in expected_columns:
        if col not in df.columns:
            missing_columns.append(col)
    if len(missing_columns) > 0:
        logger.error(f"Schema Validation Failed! Missing columns: {missing_columns}")
        return False
    logger.info("Schema Validation Successful: All expected columns are present.")
    return True

def clean_data(df):
    if df is None:
        return None
    logger.info("Starting data cleaning process.")
    initial_rows = len(df)
    df = df.drop_duplicates()
    duplicate_count = initial_rows - len(df)
    if duplicate_count > 0:
        logger.warning(f"Data Change: Dropped {duplicate_count} rows. Reason: Duplicate records detected in the dataset.")
    df['weather_main'] = df['weather_main'].str.strip().str.title()
    df['weather_description'] = df['weather_description'].str.strip().str.lower()
    df['holiday'] = df['holiday'].str.strip()
    logger.info("Standardised categorical text values.")
    try:
        df['date_time'] = pd.to_datetime(df['date_time'])
        logger.info("Parsed and validated date/time fields.")
    except Exception as e:
        logger.error(f"Date/time parsing failed: {e}")
    df['month'] = df['date_time'].dt.month
    unique_months = sorted(df['month'].dropna().unique())
    for m in unique_months:
        month_mask = df['month'] == m
        valid_temp_mask = month_mask & (df['temp'] > 0)
        valid_month_temps = df.loc[valid_temp_mask, 'temp']
        if not valid_month_temps.empty:
            monthly_median_temp = valid_month_temps.median()
            zero_temp_mask = month_mask & (df['temp'] == 0)
            zero_count = zero_temp_mask.sum()
            if zero_count > 0:
                df.loc[zero_temp_mask, 'temp'] = monthly_median_temp
                logger.warning(f"Data Change: Imputed {zero_count} rows in month {m}. Reason: Outlier temperature values showing impossible 0 Kelvin readings replaced with monthly median ({monthly_median_temp}K).")
        valid_rain_mask = month_mask & (df['rain_1h'] < 9000)
        valid_month_rain = df.loc[valid_rain_mask, 'rain_1h']
        if not valid_month_rain.empty:
            monthly_median_rain = valid_month_rain.median()
            extreme_rain_mask = month_mask & (df['rain_1h'] >= 9000)
            extreme_count = extreme_rain_mask.sum()
            if extreme_count > 0:
                df.loc[extreme_rain_mask, 'rain_1h'] = monthly_median_rain
                logger.warning(f"Data Change: Imputed {extreme_count} rows in month {m}. Reason: Outlier rainfall values exceeding physically plausible 9K mm range replaced with monthly median ({monthly_median_rain}mm).")
    df = df.drop(columns=['month'])
    return df

def clean_and_transform(df):
    if df is None:
        logger.warning("No data found to process. Stopping here.")
        return None
    logger.info("Starting data checks.")
    missing_data = df.isnull().sum()
    logger.info("Checking for missing values in each column:")
    for column, count in missing_data.items():
        if count > 0:
            logger.warning(f"Column '{column}' has {count} missing rows.")
    logger.info("Checking column data types.")
    logger.info("Extracting hour from date_time column.")
    try:
        df['hour'] = df['date_time'].dt.hour
    except Exception as e:
        logger.error(f"Could not extract hour dimension: {e}")
    df['temp_celsius'] = df['temp'] - 273.15
    logger.info("Creating traffic category column.")
    def check_volume(volume):
        if volume > 5500:
            return 'High'
        elif volume >= 2500:
            return 'Medium'
        else:
            return 'Low'
    df['traffic_category'] = df['traffic_volume'].apply(check_volume)
    logger.info("All data pipeline transformations completed successfully.")
    return df

def engineer_features(df):
    if df is None:
        return None
    
    before_rows, before_cols = df.shape
    logger.info(f"Dataset shape before feature engineering: {before_rows} rows and {before_cols} columns.")
    logger.info("Starting Task 2: Feature Engineering pipeline.")
    
    df['day_of_week'] = df['date_time'].dt.dayofweek
    
    def check_weekend(day):
        if day >= 5:
            return 1
        else:
            return 0
    df['is_weekend'] = df['day_of_week'].apply(check_weekend)
    logger.info("Appended time, day of week, and weekend indicator features.")
    
    df['hour_sin'] = np.sin(2 * np.pi * df['hour'] / 24.0)
    df['hour_cos'] = np.cos(2 * np.pi * df['hour'] / 24.0)
    logger.info("Created cyclical hour encodings using sine and cosine transformations.")
    
    weather_dummies = pd.get_dummies(df['weather_main'], prefix='weather')
    df = pd.concat([df, weather_dummies], axis=1)
    
    def check_bad_weather(row):
        if row['rain_1h'] > 0 or row['snow_1h'] > 0:
            return 1
        else:
            return 0
    df['has_precipitation'] = df.apply(check_bad_weather, axis=1)
    logger.info("Encoded categorical weather columns and created active precipitation indicators.")
    
    min_temp = df['temp'].min()
    max_temp = df['temp'].max()
    logger.debug(f"Intermediate Scaling Values: temp column absolute min={min_temp}, absolute max={max_temp}")
    if max_temp - min_temp != 0:
        df['temp_scaled'] = (df['temp'] - min_temp) / (max_temp - min_temp)
        
    min_clouds = df['clouds_all'].min()
    max_clouds = df['clouds_all'].max()
    logger.debug(f"Intermediate Scaling Values: clouds_all column absolute min={min_clouds}, absolute max={max_clouds}")
    if max_clouds - min_clouds != 0:
        df['clouds_scaled'] = (df['clouds_all'] - min_clouds) / (max_clouds - min_clouds)
    logger.info("Applied Min-Max scaling normalization to temp and clouds_all columns.")
    
    def check_congestion(volume):
        if volume > 5500:
            return 1
        else:
            return 0
    df['is_congested'] = df['traffic_volume'].apply(check_congestion)
    logger.info("Created binary traffic targets based on a 5500 vehicle congestion threshold restriction.")
    
    after_rows, after_cols = df.shape
    logger.info(f"Dataset shape after feature engineering: {after_rows} rows and {after_cols} columns.")
    
    return df

if __name__ == "__main__":
    try:
        my_file = "Metro_Interstate_Traffic_Volume.csv"
        raw_df = load_csv_data(my_file)
        if raw_df is None:
            logger.error("Pipeline cannot complete because the data file could not be loaded.")
            sys.exit(1)
        if validate_schema(raw_df):
            cleaned_df = clean_data(raw_df)
            transformed_df = clean_and_transform(cleaned_df)
            final_df = engineer_features(transformed_df)
            logger.info("Pipeline completed successfully.")
        else:
            logger.error("Pipeline stopped because schema validation failed.")
            sys.exit(1)
    except Exception as error:
        logger.error("Pipeline execution failed unexpectedly before completion.", exc_info=True)
        sys.exit(1)
```
